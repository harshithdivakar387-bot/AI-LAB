```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create monthly time series data
np.random.seed(42)

dates = pd.date_range(start='2023-01-01',
                      periods=60, freq='MS')

t = np.arange(60)

# Step 2: Generate time series characteristics
trend = 100 + 2 * t

seasonality = 100 + 20 * np.sin(2 * np.pi * t / 12)

# Illustrative long-term irregular fluctuations
cyclical = 100 + 25 * np.sin(2 * np.pi * t / 30) \
           + 10 * np.sin(2 * np.pi * t / 19)

constant_variance = 100 + np.random.normal(0, 10, 60)

# Step 3: Create DataFrame
df = pd.DataFrame({
    'Date': dates,
    'Trend': trend,
    'Seasonality': seasonality,
    'Cyclical': cyclical,
    'Constant Variance': constant_variance
})

# Step 4: Plot characteristics
fig, axes = plt.subplots(4, 1, figsize=(10, 12))

columns = ['Trend', 'Seasonality',
           'Cyclical', 'Constant Variance']

for i, col in enumerate(columns):
    axes[i].plot(df['Date'], df[col], marker='o',
                 markersize=3)
    axes[i].set_title(col)
    axes[i].set_xlabel('Date')
    axes[i].set_ylabel('Sales')
    axes[i].grid(True)

plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create time series dataset
data = {
    'Date': pd.date_range(
        start='2026-01-01', periods=14, freq='D'
    ),
    'Consumption': [100, 120, 90, 80, 140, 150, 130,
                    110, 125, 95, 85, 145, 155, 135]
}

df = pd.DataFrame(data)

# Step 2: Set Date as index
df.set_index('Date', inplace=True)

# Step 3: Extract weekday names
df['Weekday'] = df.index.day_name()

# Step 4: Group consumption by weekday
weekly_avg = df.groupby('Weekday')['Consumption'].mean()

# Step 5: Arrange weekdays in order
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday']

weekly_avg = weekly_avg.reindex(days)

print("Average Consumption by Weekday:")
print(weekly_avg)

# Step 6: Visualization
weekly_avg.plot(kind='bar', color='blue')

plt.title('Average Electricity Consumption by Weekday')
plt.xlabel('Day of the Week')
plt.ylabel('Average Consumption (GWh)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create time series dataset
data = {
    'Date': ['2026-01-01', '2026-01-02', '2026-01-03',
             '2026-02-01', '2026-02-02', '2026-02-03',
             '2026-03-01', '2026-03-02', '2026-03-03'],
    'Electricity': [100, 120, 140, 150, 180, 210,
                    200, 220, 240]
}

df = pd.DataFrame(data)

# Step 2: Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Step 3: Group data by month
monthly_data = df.groupby(
    df['Date'].dt.to_period('M')
)['Electricity'].agg(['sum', 'mean'])

# Step 4: Rename columns
monthly_data.columns = ['Total', 'Average']

print(monthly_data)

# Step 5: Plot monthly total consumption
monthly_data['Total'].plot(kind='bar', marker='o')

plt.title('Monthly Electricity Consumption')
plt.xlabel('Month')
plt.ylabel('Electricity (kWh)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Resampling is the process of changing the frequency of time series data from one time interval to another.
#A weather station records daily temperatures. Convert daily temperature data into weekly average temperature using resampling.
import pandas as pd

data = {
    'Date': pd.date_range('2026-01-01', periods=14),
    'Temperature': [25, 26, 27, 28, 29, 30, 31,
                    24, 25, 26, 27, 28, 29, 30]
}

df = pd.DataFrame(data)

df.set_index('Date', inplace=True)

# Resample into 7-day intervals
weekly = df.resample('7D').mean()

print(weekly)
```

```python
#Grouping time series data is the process of organizing observations based on time intervals such as days, months, or years and performing calculations like sum, mean, minimum, and maximum.

import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create traffic data
data = {
    'Time': ['08:00', '08:15', '09:00',
             '09:15', '10:00', '10:15'],
    'Vehicles': [100, 150, 200, 250, 120, 180]
}

df = pd.DataFrame(data)

# Step 2: Convert time into datetime
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M')

# Step 3: Extract hour
df['Hour'] = df['Time'].dt.hour

# Step 4: Group by hour
traffic = df.groupby('Hour')['Vehicles'].sum()

print(traffic)

# Step 5: Plot graph
traffic.plot(kind='bar')

plt.title('Hourly Traffic Analysis')
plt.xlabel('Hour')
plt.ylabel('Total Vehicles')
plt.xticks(rotation=0)
plt.show()
```

```python
import pandas as pd
#df.resample('MS').mean()
#resample() – Groups observations into time intervals.
# 'MS' – Monthly frequency, labeled by the start of each month.
# mean() – Calculates the average of the observations in each month.
# Step 1: Create dataset
data = {
    'Date': ['2026-01-01', '2026-01-02',
             '2026-02-01', '2026-02-02'],
    'Consumption': [100, 200, 300, 400]
}

df = pd.DataFrame(data)

# Step 2: Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Step 3: Set Date as index
df.set_index('Date', inplace=True)

# Step 4: Resample monthly
monthly_avg = df.resample('MS').mean()

print(monthly_avg)
```

```python
#Group-wise transformation is the process of applying a calculation to each group of data and assigning the calculated results back to the original rows.
import pandas as pd

# Step 1: Create dataset
data = {
    'Date': ['2026-01-01', '2026-01-02',
             '2026-01-03', '2026-02-01',
             '2026-02-02', '2026-02-03'],
    'Sales': [100, 200, 300, 400, 500, 600]
}

df = pd.DataFrame(data)

# Step 2: Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Step 3: Extract month
df['Month'] = df['Date'].dt.to_period('M')

# Step 4: Group-wise transformation
df['Monthly_Avg'] = df.groupby('Month')['Sales'].transform('mean')

# Step 5: Calculate difference from monthly average
df['Difference'] = df['Sales'] - df['Monthly_Avg']

print(df)
```

```python
import pandas as pd

# Step 1: Create dataset
df = pd.DataFrame({
    'Month': ['January', 'January', 'January',
              'February', 'February', 'February'],
    'Sales': [100, 200, 300, 400, 500, 600]
})

# Step 2: Group by month and calculate total sales
result = df.groupby('Month')['Sales'].sum().reset_index()

# Step 3: Rename Sales column
result = result.rename(columns={'Sales': 'Total_Sales'})

print(result)
```
