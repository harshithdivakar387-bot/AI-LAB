```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Step 1: Create dataset
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 9],
    'Marks': [40, 50, 55, 65, 70, 80, 85, 95],
    'Department': ['CSE', 'ECE', 'CSE', 'ECE',
                   'CSE', 'ECE', 'CSE', 'ECE'],
    'Attendance': [60, 65, 70, 75, 80, 85, 90, 95],
    'Gender': ['M', 'F', 'F', 'M',
               'M', 'F', 'F', 'M']
}

df = pd.DataFrame(data)

# Step 2: Mapping data onto aesthetics
sns.scatterplot(
    data=df,
    x='Study_Hours',       # Position on X-axis
    y='Marks',             # Position on Y-axis
    hue='Department',  # Colour aesthetic hue is an aesthetic mapping that uses colour to distinguish data values or categories.
    size='Attendance',     # Size aesthetic
    style='Gender',        # Shape aesthetic
    sizes=(50, 400),
    alpha=0.8
)

# Step 3: Set position scales Controls the range of X-axis values
plt.ylim()	#Controls the range of Y-axis values
plt.xlim(0, 10)
plt.ylim(0, 100)

# Step 4: Add labels
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Mapping Data onto Aesthetics')

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Create dataset
data = {
    'Branch': ['A', 'B', 'C', 'D'],
    'Sales': [10, 100, 1000, 10000]
}

df = pd.DataFrame(data)

# Step 2: Apply log transformation
df['Log_Sales'] = np.log10(df['Sales'])

print(df)

# Step 3: Plot original sales
plt.figure(figsize=(8, 4))
plt.bar(df['Branch'], df['Sales'])
plt.title('Original Sales')
plt.xlabel('Branch')
plt.ylabel('Sales')
plt.show()

# Step 4: Plot log-transformed sales
plt.figure(figsize=(8, 4))
plt.bar(df['Branch'], df['Log_Sales'])
plt.title('Log Transformed Sales')
plt.xlabel('Branch')
plt.ylabel('Log10(Sales)')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

sales = [1, 4, 9, 16, 25]
shops = ['A', 'B', 'C', 'D', 'E']

# Square root transformation
sqrt_sales = np.sqrt(sales)

print("Original Sales:", sales)
print("Square Root Sales:", sqrt_sales)

plt.bar(shops, sqrt_sales)

plt.xlabel('Shops')
plt.ylabel('Square Root of Sales')
plt.title('Square Root Transformation')

plt.show()
```
